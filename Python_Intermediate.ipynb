## 1. OOP
Object Oriented Programming
Python is object oriented programming language since it use class and object to create something meaningful. 
Class is the blue print for the object

```python
# Let's create a class and object in Python
class Student(object):
    def __init__(self, name, age): # This is called initialized object constructor
        self.name = name
        self.age = age

    def introduce(self): # This is method it give us more details about class
        return f"Hello, my name is {self.name} and I am {self.age} years old"

gatduel = Student("Gatduel", 27) # This is what we called object
print(gatduel.introduce())
```

```python
charles = Student("Charles", 25) # This is also another object
print(charles.introduce())
```

## 3. try-except
In Python we have 3 types of errors
<ol>
    <li>Syntax errors</li>
    <li>Logic errors</li>
    <li>Exception errors</li>
</ol>
Here we will focus on how to handle exception in Python. Exception error occurs when we try to do the impossible in Python. If we try to do what is not in Python like opening a file without a real file path, use wrong data type to perform operation or dividing a number by zero. They are very many so you need to do research about them. Now try-except error prevent your code from crushing into exceptional errors, instead it provide a message that give the user clue that they have a mistake

```python
# Here is the code without a try-except
name = input("What is your name? ") # accept only strings, if you try to put different data types the code will crush
age = int(input("How old are you? ")) # accept only integers, if you use non-integers the code will crush

print(f"My name is {name}")
print(f"I am {age} years old")
```

```python
# Here is the same code with try-accept, it prevent the whole program from code crush
try:
    student_name = input("What is your name? ") # The print function below the variable will prompt the user that this field contain string only
    
except:
    print("This field accept the string or text-only")
try:
    age = int(input("How old are you? "))

except: 
    print("The second line accept numbers only")
    
print(f"My name is {student_name}")
print(f"I am {age} years old")
```

## To scrape data you need to import the libraries

```python
# Import libraries
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

<p><b>The header variable prevents your scraping from being blocked, and the URL variable contains the URL 
of the websites you want to scrape. Scraping is for automation; it is like sending a robot to the website to fetch some data for you</b></p>

<p>To start practicing web scraping visit this url <a href = "https://www.scrapethissite.com/pages/" title = "Web scraping site" target = "_blank">Web Scraping site</a>
</p>

```python
# Header and URL variables: I am scraping countries of the world
Header = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"}
page_url = "https://www.scrapethissite.com/pages/simple/"
```

```python
# Work with the requests library to download the structure of the website content
page = requests.get(page_url, headers = Header)
```

```python
# Check the status of your website; if it is 200, it means you can scrape the website
page
# Since our status is 200 now, we can move on and scrape the data
```

```python
# Use BeautifulSoup to parse the HTML contents
soup = BeautifulSoup(page.text, "html.parser")
#soup
```

```python
# Now scrape the names of the countries
Title = soup.find_all("h3")
#Title
```

```python
country_names = [headers.text.strip() for headers in Title]
#country_names
```

```python
# Now let's scrape the page's capital cities
capital = soup.find_all("span", class_ = "country-capital")
```

```python
country_capital = [city.text.strip() for city in capital]
#country_capital
```

```python
# Let's scrape the population
ppl = soup.find_all("span", class_ = "country-population")
```

```python
country_popula = [popu.text.strip() for popu in ppl]
```

```python
# Lastly, let's scrape the area in km2 of the countries
country_area = soup.find_all("span", class_ = "country-area")
```

```python
# Iterate through the data and add it to the variable
Area = [area.text.strip() for area in country_area]
Area
```

## Pandas territory

```python
# Now let's create our data structure using a Pandas DataFrame
df = pd.DataFrame()
df["Country Names"] = country_names
df["Capital Cities"] = country_capital
df["Country Populations"] = country_popula
df["Country Area in km2"] = Area
```
